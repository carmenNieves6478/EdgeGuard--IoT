# ⚡ EdgeGuard-IoT: Step 03 - INT8 Post-Training Quantization & Explainable AI (SHAP)
**Objective:** Export PyTorch model to ONNX, apply INT8 dynamic quantization for Edge AI deployment, evaluate compression & latency speedups, and calculate SHAP feature importances.

```python
import os, json, numpy as np, pandas as pd
import matplotlib.pyplot as plt
from IPython.display import Image, display

size_fp32 = os.path.getsize('../models/vae_mlp.onnx') / 1024
size_int8 = os.path.getsize('../models/vae_mlp_quantized.onnx') / 1024

print('====================================================')
print('      MLOps FOOTPRINT DE COMPRESIÓN EN EL BORDE      ')
print('====================================================')
print(f" - Modelo ONNX Float32:  {size_fp32:.2f} KB")
print(f" - Modelo ONNX INT8:     {size_int8:.2f} KB")
print(f" - Tasa de Compresión:  {((size_fp32 - size_int8)/size_fp32)*100:.2f}%")
print('====================================================')
```
